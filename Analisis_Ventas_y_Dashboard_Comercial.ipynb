## 2. Importar librerías

````python
import pandas as pd
import matplotlib.pyplot as plt
````

## 3. Crear datos de ventas

````python
datos = {
    "Producto": [
        "Laptop",
        "Mouse",
        "Teclado",
        "Monitor",
        "Laptop",
        "Mouse",
        "Monitor",
        "Teclado"
    ],
    "Ventas": [3500, 80, 150, 900, 4200, 100, 950, 180],
    "Cantidad": [1, 2, 1, 1, 1, 3, 1, 2]
}

df = pd.DataFrame(datos)
df
````

## 4. Obtener información general

````python
print(df.info())
````

## 5. Ventas totales

````python
ventas_totales = df["Ventas"].sum()
print("Ventas Totales:", ventas_totales)
````

## 6. Producto más vendido

````python
producto_mas_vendido = df.groupby("Producto")["Cantidad"].sum()
print(producto_mas_vendido)
````

## 7. Crear gráfico comercial

````python
producto_mas_vendido.plot(
    kind='bar',
    color='skyblue'
)
plt.title("Cantidad vendida por producto")
plt.xlabel("Producto")
plt.ylabel("Cantidad")
plt.grid(True)
plt.show()
````

## 8. Producto con mayores ingresos

````python
ingresos = df.groupby("Producto")["Ventas"].sum()
print(ingresos)
````

## 9. Gráfico de ingresos

````python
ingresos.plot(
    kind='pie',
    autopct='%1.1f%%'
)
plt.title("Participación de ingresos")
plt.ylabel("")
plt.show()
````

## 3. Importar librerías

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

## 4. Crear base de datos comercial

````python
datos = {
    "Fecha": [
        "2026-01-10",
        "2026-01-11",
        "2026-01-12",
        "2026-01-13",
        "2026-01-14",
        "2026-01-15",
        "2026-01-16",
        "2026-01-17"
    ],
    "Producto": [
        "Laptop",
        "Mouse",
        "Teclado",
        "Monitor",
        "Laptop",
        "Mouse",
        "Monitor",
        "Teclado"
    ],
    "Vendedor": [
        "Ana",
        "Luis",
        "Ana",
        "Carlos",
        "Luis",
        "Ana",
        "Carlos",
        "Luis"
    ],
    "Ventas": [
        3500,
        80,
        150,
        900,
        4200,
        100,
        950,
        180
    ],
    "Cantidad": [
        1,
        2,
        1,
        1,
        1,
        3,
        1,
        2
    ]
}

df = pd.DataFrame(datos)
df
````

## 5. Convertir fechas

````python
df["Fecha"] = pd.to_datetime(df["Fecha"])
df.info()
````

## 6. KPI principal

````python
ventas_totales = df["Ventas"].sum()
print("Ventas Totales:", ventas_totales)
````

## 7. KPI por producto

````python
ventas_producto = df.groupby("Producto")["Ventas"].sum()
print(ventas_producto)
````

## 8. Gráfico profesional

````python
plt.figure(figsize=(8,5))
sns.barplot(
    x=ventas_producto.index,
    y=ventas_producto.values,
    palette="Blues_d"
)
plt.title("Ventas por Producto")
plt.xlabel("Producto")
plt.ylabel("Ingresos")
plt.show()
````

## 9. Análisis de vendedores

````python
ventas_vendedor = df.groupby("Vendedor")["Ventas"].sum()
print(ventas_vendedor)
````

## 10. Top vendedor

````python
top_vendedor = ventas_vendedor.idxmax()
print("Mejor vendedor:", top_vendedor)
````

## 11. Gráfico circular

````python
ventas_vendedor.plot(
    kind='pie',
    autopct='%1.1f%%',
    figsize=(6,6)
)
plt.title("Participación por vendedor")
plt.ylabel("")
plt.show()
````

## 12. Análisis temporal

````python
ventas_fecha = df.groupby("Fecha")["Ventas"].sum()
print(ventas_fecha)
````

## 13. Tendencia de ventas

````python
plt.figure(figsize=(10,5))
plt.plot(
    ventas_fecha.index,
    ventas_fecha.values,
    marker='o'
)
plt.title("Tendencia de Ventas")
plt.xlabel("Fecha")
plt.ylabel("Ventas")
plt.grid(True)
plt.show()
````
